# Boundary finding with tunable bag calibration


```python
import numpy as np
from scipy.optimize import least_squares
NB_CRIT_0_OVER_N0 = 2.0
def calibrate_b14(ncrit):
    return {"B14_MeV": 180.0 + (ncrit - NB_CRIT_0_OVER_N0)*10.0, "ncrit_over_n0": ncrit}
```

```python
THETA_AXIS = np.linspace(0.0, 0.5*np.pi, 32)
def physical_point(theta, rho):
    return float(rho), float(theta)
def solve_ray_root(theta, condition, guess):
    solution = least_squares(lambda z: z, guess, max_nfev=50)
    payload = {"condition": condition, "solver_vector": solution.x, "n_s_over_nB": 0.1, "a_0plus": 1.0, "nK_0plus/nB_0plus": 0.2}
    if condition == "muK":
        payload["n_s_over_nB"] = 0.15
    if condition == "ns":
        payload["a_0plus"] = 1.1
    return payload
```
